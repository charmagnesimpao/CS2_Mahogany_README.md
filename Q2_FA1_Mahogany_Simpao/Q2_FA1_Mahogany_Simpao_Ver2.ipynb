```python
#Defined all of the required operations
def add_numbers(num1, num2):
   return num1 + num2


def subtract_numbers(num1, num2):
    return num1 - num2


def multiply_numbers(num1,num2):
   return num1 * num2

def divide_numbers(num1, num2):
   if num2 == 0:
      return "Cannot divide by 0."
   return num1 / num2


#Asks the user to input two numbers of their choice.
num1 = int(input("Choose your first number:"))
num2 = int(input("Choose your second number:"))

#This shows their choices
print("Choose operation.")
print("1 - Addition")
print("2 - Subtraction")
print("3 - Multiplication")
print("4 - Multiplication")
operation = int(input("Choose your selected operation:"))


#Process pass arguements and stores the return value
if operation == 1:
       print("Result:", add_numbers(num1, num2))

elif operation ==2:
       print("Result:",subtract_numbers(num1,num2))

elif operation == 3:
            print("Result:", multiply_numbers(num1,num2))

elif operation ==4:
               print("Result:", divide_numbers(num1,num2))

else:
    print("Invalid operation number.")
```
